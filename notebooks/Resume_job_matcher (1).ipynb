```python

import pandas as pd
import re

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries imported successfully!")
```

```python

resume_text = """
BSc graduate with knowledge of Python, SQL, Pandas,
Excel, data cleaning, data analysis, and visualization.
Completed projects using Python and Power BI.
"""

job_description = """
We are looking for a data analyst with skills in Python,
SQL, Excel, Power BI, data cleaning, data visualization,
and reporting. Knowledge of Pandas is preferred.
"""

print("Resume text loaded:", bool(resume_text.strip()))
print("Job description loaded:", bool(job_description.strip()))
```

```python

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z0-9+#.\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

cleaned_resume = clean_text(resume_text)
cleaned_job = clean_text(job_description)

print("Cleaned Resume:")
print(cleaned_resume)

print("\nCleaned Job Description:")
print(cleaned_job)
```

```python

# Convert text into numerical features
vectorizer = TfidfVectorizer(stop_words="english")

tfidf_matrix = vectorizer.fit_transform([
    cleaned_resume,
    cleaned_job
])

# Calculate cosine similarity
similarity = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

match_percentage = similarity * 100

print("Resume–Job Similarity:",
      round(match_percentage, 2), "%")
```

```python

# Define the skills we want to check
skills = [
    "python",
    "sql",
    "excel",
    "power bi",
    "pandas",
    "data cleaning",
    "data analysis",
    "data visualization",
    "reporting",
    "machine learning"
]

# Find skills mentioned in each text
resume_skills = [
    skill for skill in skills
    if skill in cleaned_resume
]

job_skills = [
    skill for skill in skills
    if skill in cleaned_job
]

# Compare skills
matched_skills = [
    skill for skill in job_skills
    if skill in resume_skills
]

missing_skills = [
    skill for skill in job_skills
    if skill not in resume_skills
]

print("Skills found in resume:", resume_skills)
print("Skills requested by job:", job_skills)
print("Matching skills:", matched_skills)
print("Potential skill gaps:", missing_skills)
```

```python

print("=" * 45)
print("RESUME–JOB MATCH REPORT")
print("=" * 45)

print(f"\nText similarity: {match_percentage:.2f}%")

print("\nMatching skills:")
if matched_skills:
    for skill in matched_skills:
        print("-", skill)
else:
    print("No matching skills detected.")

print("\nPotential skill gaps:")
if missing_skills:
    for skill in missing_skills:
        print("-", skill)
else:
    print("No gaps detected in the selected skill list.")

print("\nNote: Similarity is not a hiring probability.")
print("Review the resume and job description manually.")
```

```python

# New test resume
test_resume = """
Graduate with Python, SQL, Excel, Pandas,
data analysis, and data cleaning skills.
Experience creating reports and dashboards.
"""

# New job description
test_job = """
Hiring a junior data analyst with Python,
SQL, Excel, Power BI, data analysis,
data visualization, and reporting skills.
"""

# Clean both texts
test_resume_clean = clean_text(test_resume)
test_job_clean = clean_text(test_job)

# Calculate similarity
test_vectors = vectorizer.fit_transform([
    test_resume_clean,
    test_job_clean
])

test_similarity = cosine_similarity(
    test_vectors[0:1],
    test_vectors[1:2]
)[0][0] * 100

print("Test similarity:", round(test_similarity, 2), "%")
```

```python

test_resume_skills = [
    skill for skill in skills
    if skill in test_resume_clean
]

test_job_skills = [
    skill for skill in skills
    if skill in test_job_clean
]

test_matched = [
    skill for skill in test_job_skills
    if skill in test_resume_skills
]

test_gaps = [
    skill for skill in test_job_skills
    if skill not in test_resume_skills
]

print("Matching skills:", test_matched)
print("Potential skill gaps:", test_gaps)
```
